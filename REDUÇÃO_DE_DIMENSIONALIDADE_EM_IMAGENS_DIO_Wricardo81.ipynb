```python
# Install dependencies
!pip install opencv-python matplotlib --quiet

# Imports
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload da imagem
print("📤 Faça upload de uma imagem
uploaded = files.upload()

# Captura o nome do arquivo
image_path = next(iter(uploaded))

# Carrega e converte para RGB
img_colorida = cv2.imread(image_path)
img_colorida = cv2.cvtColor(img_colorida, cv2.COLOR_BGR2RGB)

# Converte para escala de cinza
img_cinza = cv2.cvtColor(img_colorida, cv2.COLOR_RGB2GRAY)

# Binarização com threshold
limiar = 127
_, img_binaria = cv2.threshold(img_cinza, limiar, 255, cv2.THRESH_BINARY)

# Plota os resultados
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title("Imagem Colorida")
plt.imshow(img_colorida)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Escala de Cinza")
plt.imshow(img_cinza, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Imagem Binarizada")
plt.imshow(img_binaria, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
